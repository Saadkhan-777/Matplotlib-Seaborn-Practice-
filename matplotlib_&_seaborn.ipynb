## Bar Plots

```python
years = [1990, 1992, 1994, 1996, 1998, 2000, 2003, 2005, 2007, 2010]
runs = [500, 700, 1100, 1500, 1800, 1200, 1700, 1300, 900, 1500]

plt.bar(years, runs)
plt.title("Sachin tendulkar yearly runs")
plt.xlabel("years")
plt.ylabel("runs")
plt.show()
```

```python
x = np.arange(len(years))  # index position
print(x)
```

```python
sachin = [500, 700, 1100, 1500, 1800, 1200, 1700, 1300, 900, 1500]
sehwag = [0, 300, 800, 1200, 1500, 1700, 1600, 1400, 1000, 0]
kohli = [0, 0, 500, 800, 1100, 1300, 1500, 1800, 1900, 2100]

x = np.arange(len(years))     # index position
width = 0.25       
```

The value of width is from 0 to 1 (by default it is 0.8)

```python
plt.bar(x - width, sachin, width=width, label="Sachin Tendulkar" )
plt.bar(x,         sehwag, width=width, label="V-Sehwag" )
plt.bar(x + width, kohli,  width=width, label="Virat Kohli" )

plt.title("Comparision of Players")
plt.xlabel("Years")
plt.ylabel("Runs")
plt.legend()            # to show small label window on top of plot       
plt.xticks(x, years)    # to actual years instead of 0,1,2,3...
plt.tight_layout()      
plt.show()
```

## Pie Charts

```python
Players = ["Sachin", "Sehwag", "Kohli", "Yuraj"]
runs = [18000, 8000, 12000, 9500]

plt.title("Runs of Batters")
plt.pie(runs, labels=Players)
plt.tight_layout()
plt.show()
```

```python
Players = ["Sachin", "Sehwag", "Kohli", "Yuraj"]
runs = [18000, 8000, 12000, 9500]

plt.style.use("ggplot")
plt.title("Runs of Batters")
plt.pie(runs, labels=Players)
plt.show()
```

```python
# Custom Colors

Players = ["Sachin", "Sehwag", "Kohli", "Yuraj"]
runs = [18000, 8000, 12000, 9500]

colors = ["red", "blue", "black", "orange"]     # can also use haxcolor 
plt.title("Runs of Batters")
plt.pie(runs, labels=Players, colors=colors)
plt.show()
```

```python
# Wedgeprops

Players = ["Sachin", "Sehwag", "Kohli", "Yuraj"]
runs = [18000, 8000, 12000, 9500]

colors = ["red", "blue", "pink", "orange"]   
plt.title("Runs of Batters")
plt.pie(runs, labels=Players, colors=colors, 
        wedgeprops={'edgecolor': 'black',   'linewidth': 3,   'linestyle':'--'}
       )
plt.show()
```

```python
# fill=True or false

Players = ["Sachin", "Sehwag", "Kohli", "Yuraj"]
runs = [18000, 8000, 12000, 9500]

colors = ["red", "blue", "pink", "orange"]   
plt.title("Runs of Batters")
plt.pie(runs, labels=Players, colors=colors, 
        wedgeprops={'edgecolor': 'black',   'linewidth': 2,   'linestyle':'--', 'fill':False}
       )
plt.show()
```

```python
# Explode argument

explode = [0.1, 0, 0, 0]    # sachin exploded 0.1 

plt.title("Explode pie example")
plt.pie(runs, labels=Players, colors=colors, explode=explode)
plt.show()
```

```python
# Shadow and Angles

plt.pie(runs, labels=Players, colors=colors, explode=explode, shadow=True, startangle=140)
plt.show()
```

```python
# autopct      very import.
plt.pie(runs, labels=Players, colors=colors, explode=explode, startangle=140, autopct="%1.1f%%")
plt.show()
```

## Stack Plots in Matplotlib

```python
activities = ["Studying", "Playing", "Watch TV", "Sleeping"]
time_spent = [3, 2, 2, 5]

colors = ['blue', 'red', 'yellow', 'pink']
plt.figure(figsize=(4, 4))                                     # size of Pie
plt.pie(time_spent, labels=activities, colors=colors, autopct="%1.1f%%", startangle=90)
plt.title("Student daily_major_activities")
plt.axis("equal")                                              # equal circle pie
plt.show() 
```

This plot couldnot tell us how many hours he spend on sleeping from total

```python
days = [1, 2, 3, 4, 5, 6, 7]
study = [3, 4, 3, 5, 4, 7, 4]
play = [2, 2, 1, 1, 2, 3, 3]
watch_TV = [2, 3, 2, 1, 1, 3, 1]
sleep = [6, 4, 7, 8, 7, 6, 5]

labels = ['study', 'play', 'TV', 'Sleep' ]
colors = ['blue', 'red', 'yellow', 'pink']
plt.figure(figsize=(4, 4))
plt.stackplot(days, study, play, watch_TV, sleep, labels=labels, colors=colors, alpha=0.4 )  # alpha = opacity
plt.legend()
plt.show()

# give title, xlabel, ylabel, grid ..
```

```python
# Adding Axvline on Average age

plt.hist(ages, bins=bin, edgecolor="black")

plt.axvline(np.mean(ages), color="black" , linewidth=4, linestyle="--", label="AVG Age")
plt.xlabel("Age Group")
plt.ylabel("No of Viewers")
plt.title("Age distribution of YT Viewers")
plt.grid(True, linestyle="--")
plt.show()
```

```python
# Adding Axvline on random age

plt.hist(ages, bins=bin, edgecolor="black")

plt.axvline(15, color="black" , linewidth=4, linestyle="--", label="AVG Age")
plt.xlabel("Age Group")
plt.ylabel("No of Viewers")
plt.title("Age distribution of YT Viewers")
plt.grid(True, linestyle="--")
plt.show()
```

## Scatter Plots in Matplotlib

```python
study_hours = [1, 2, 3, 4, 5, 6, 7, 8, 9]
exam_scores = [40, 45, 50, 55, 60, 65, 75, 85, 90]

plt.scatter(study_hours, exam_scores)
plt.title("Study Hours vs Exam score")
plt.xlabel("Study Hours")
plt.ylabel("Exam Score")
plt.grid(True)
plt.show()
```

```python
study_hours = [1, 2, 3, 4, 5, 6, 7, 8, 9]
exam_scores = [40, 45, 90, 55, 60, 25, 75, 85, 90]

plt.scatter(study_hours, exam_scores)
plt.title("Study Hours vs Exam score")
plt.xlabel("Study Hours")
plt.ylabel("Exam Score")
plt.grid(True)
plt.show()
```

```python
# Adding colors and sizes

colors = ['red' if score <50 else 'green' for score in exam_scores]
sizes = [score * 2 for score in exam_scores]

plt.scatter(study_hours, exam_scores, c=colors, s=sizes)
plt.title("Study Hours vs Exam score")
plt.xlabel("Study Hours")
plt.ylabel("Exam Score")
plt.grid(True)
plt.show()
```

```python
# using cmaps   -- U can aslo search on internet

# colors = ['red' if score <50 else 'green' for score in exam_scores]
# sizes = [score * 2 for score in exam_scores]

plt.scatter(study_hours, exam_scores, c=exam_scores, cmap='viridis')
plt.colorbar(label='Score')

plt.title("Study Hours vs Exam score")
plt.xlabel("Study Hours")
plt.ylabel("Exam Score")
plt.grid(True)
plt.show()
```

```python
# Adding Annotations

for i in range(len(study_hours)):
    plt.annotate(f' St{i+1}' , (study_hours[i] , exam_scores[i] -2 ) )

plt.scatter(study_hours, exam_scores, c=exam_scores, cmap='viridis')
plt.colorbar(label='Score')

plt.title("Study Hours vs Exam score")
plt.xlabel("Study Hours")
plt.ylabel("Exam Score")
plt.grid(True)
plt.show()
```
