### Multiple linePlot Im One Plot

```python
years = [1990, 1992, 1994, 1996, 1998, 2000, 2003, 2005, 2007, 2010]

kohli = [0, 0, 500, 800, 1100, 1300, 1500, 1800, 1900, 2100]
sehwag = [0, 300, 800, 1200, 1500, 1700, 1600, 1400, 1000, 0]

plt.plot(years, kohli)
plt.plot(years, sehwag)

plt.title("Performance Comparision")
plt.xlabel("Years")
plt.ylabel("Runs Scored")
plt.show()
```

```python
# From the upper table we can't get insights
# So we use legend function

years = [1990, 1992, 1994, 1996, 1998, 2000, 2003, 2005, 2007, 2010]

kohli = [0, 0, 500, 800, 1100, 1300, 1500, 1800, 1900, 2100]
sehwag = [0, 300, 800, 1200, 1500, 1700, 1600, 1400, 1000, 0]

plt.plot(years, kohli, label = "Virat Kohli")
plt.plot(years, sehwag, label = "Virender Sehwag")

plt.title("Performance Comparision")
plt.xlabel("Years")
plt.ylabel("Runs Scored")
plt.legend() 
plt.show()

# NOTE  we can also try this combinition -- but not recommended
# plt.plot(years, kohli)
# plt.plot(years, sehwag)
# plt.legend(["Virat Kohli", "Virender Sehwag"]) 
```

```python
# Formate string

plt.plot(years, kohli,"ro--", label = "Virat Kohli")          # red circle with dash line
plt.plot(years, sehwag,"g^:", label = "Virender Sehwag")      # green traingle dotted line
plt.legend()
plt.show()
```

```python
# color and line Style Argument

plt.plot(years, kohli,   color="red", linestyle="--",   label = "Virat Kohli")          
plt.plot(years, sehwag,  color="blue", linestyle=":",   label = "Virender Sehwag")     
plt.legend()
plt.show()
```

```python
plt.plot(years, kohli,   color="red", linestyle="--",  linewidth=3 , label = "Virat Kohli")          
plt.plot(years, sehwag,  color="blue", linestyle=":",   label = "Virender Sehwag") 
plt.grid(True)
plt.legend()
plt.show()
```

## Bar Plots

```python
years = [1990, 1992, 1994, 1996, 1998, 2000, 2003, 2005, 2007, 2010]
runs = [500, 700, 1100, 1500, 1800, 1200, 1700, 1300, 900, 1500]

plt.bar(years, runs)
plt.title("Sachin tendulkar yearly runs")
plt.xlabel("years")
plt.ylabel("runs")
plt.show()
```

```python
x = np.arange(len(years))  # index position
print(x)
```

```python
sachin = [500, 700, 1100, 1500, 1800, 1200, 1700, 1300, 900, 1500]
sehwag = [0, 300, 800, 1200, 1500, 1700, 1600, 1400, 1000, 0]
kohli = [0, 0, 500, 800, 1100, 1300, 1500, 1800, 1900, 2100]

x = np.arange(len(years))     # index position
width = 0.25       
```

The value of width is from 0 to 1 (by default it is 0.8)

```python
plt.bar(x - width, sachin, width=width, label="Sachin Tendulkar" )
plt.bar(x,         sehwag, width=width, label="V-Sehwag" )
plt.bar(x + width, kohli,  width=width, label="Virat Kohli" )

plt.title("Comparision of Players")
plt.xlabel("Years")
plt.ylabel("Runs")
plt.legend()            # to show small label window on top of plot       
plt.xticks(x, years)    # to actual years instead of 0,1,2,3...
plt.tight_layout()      
plt.show()
```

### Horizotal bar charts : barh()

```python
Players = ["Sachin", "Sehwag", "Kohli", "Yuraj"]
runs_5years = [ 500+700+1100+1500+1800,  0+200+900+1400+1600,  0+0+500+800+1100,  300+600+800+1100+900 ]

plt.barh(Players, runs_5years, color="indigo",)
plt.title("Performance of 5 years")
plt.xlabel("Total runs in 5year")
plt.ylabel("Players names")
plt.show()
```

## Pie Charts

```python
Players = ["Sachin", "Sehwag", "Kohli", "Yuraj"]
runs = [18000, 8000, 12000, 9500]

plt.title("Runs of Batters")
plt.pie(runs, labels=Players)
plt.tight_layout()
plt.show()
```

```python
Players = ["Sachin", "Sehwag", "Kohli", "Yuraj"]
runs = [18000, 8000, 12000, 9500]

plt.style.use("ggplot")
plt.title("Runs of Batters")
plt.pie(runs, labels=Players)
plt.show()
```

```python
# Custom Colors

Players = ["Sachin", "Sehwag", "Kohli", "Yuraj"]
runs = [18000, 8000, 12000, 9500]

colors = ["red", "blue", "black", "orange"]     # can also use haxcolor 
plt.title("Runs of Batters")
plt.pie(runs, labels=Players, colors=colors)
plt.show()
```

```python
# Wedgeprops

Players = ["Sachin", "Sehwag", "Kohli", "Yuraj"]
runs = [18000, 8000, 12000, 9500]

colors = ["red", "blue", "pink", "orange"]   
plt.title("Runs of Batters")
plt.pie(runs, labels=Players, colors=colors, 
        wedgeprops={'edgecolor': 'black',   'linewidth': 3,   'linestyle':'--'}
       )
plt.show()
```

```python
# fill=True or false

Players = ["Sachin", "Sehwag", "Kohli", "Yuraj"]
runs = [18000, 8000, 12000, 9500]

colors = ["red", "blue", "pink", "orange"]   
plt.title("Runs of Batters")
plt.pie(runs, labels=Players, colors=colors, 
        wedgeprops={'edgecolor': 'black',   'linewidth': 2,   'linestyle':'--', 'fill':False}
       )
plt.show()
```

```python
# Explode argument

explode = [0.1, 0, 0, 0]    # sachin exploded 0.1 

plt.title("Explode pie example")
plt.pie(runs, labels=Players, colors=colors, explode=explode)
plt.show()
```

```python
# Shadow and Angles

plt.pie(runs, labels=Players, colors=colors, explode=explode, shadow=True, startangle=140)
plt.show()
```

```python
# autopct      very import.
plt.pie(runs, labels=Players, colors=colors, explode=explode, startangle=140, autopct="%1.1f%%")
plt.show()
```
